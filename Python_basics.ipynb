1. What is Python, and why is it popular?
   - Python is a high level programming language used for data analysis, web development and automation. It is popular because it is easy to learn and use, versatile, has powerful libraries (like Pandas, Numpy), offers strong community support, and is free and open source.

2. What is an interpreter in Python?
   - An interpreter in Python is a program that executes code line by line, converting it into machine-readable instructions. Unlike a compiler, it runs the code directly without creating a separate executable file, making debugging easier and faster.
  

3. What are pre-defined keywords in Python?
   - The pre-defined keywords are reserved words that have a special meaning and are used for a specific purpose in Python.
   examples: if, else, for, while, etc

4. Can keywords be used as variable names?
   - No, keywords cannot be used as variable names because they are reserved words with special meanings in Python.


5. What is mutability in Python?
   - Mutability in Python refers to whether an object value can be changed after it is created.


6. Why are lists mutable, but tuples are immutable?
   - Lists are mutable because they are designed for flexibility, where as tuples are immutable because they are meant to be fixed.

7. What is the difference between “==” and “is” operators in Python?
   - "==" is an equality operator. It checks whether the valus of two objects are the same. "is" is an identity operator. It checks whether two variables refer to same object in memory.

8. What are logical operators in Python?
   - Logical operators are used to combine or modify conditions and return in boolean result (True or False). Used in making decisions based on multiple conditions.
---



9. What is type casting in Python?
   - Type casting is the process of converting one data type into another.

10. What is the difference between implicit and explicit type casting?
    - Implicit Type casting is automatically done by Python. Explicit type casting is manually done by the programmer.

11. What is the purpose of conditional statements in Python?
    - The purpose of conditional statements in Python is to make decisions in a program. They allow the program to execute different blocks of code based on certain conditions.

12. How does the elif statement work?
    - It works between 'if' and 'else'. If the 'if' condition is false, Python checks the elif condition. If that is also false, it moves to next elif and so on. Once the condition is true, its block executes and the rest are skipped.

13. What is the difference between for and while loops?
    - A for loop is used when the number of iterations is known. It runs over a sequence(like a list, string or range). A while loop is used when the number of iterations is not known. It runs as long as given condition is true.

14. Describe a scenario where a while loop is more suitable than a for loop?
    - A while loop is more suitable in situations where the number of iterations is not known and the loop must continue until the condition is met. Example: where system ask the user to input a correct password.

```python
#6. Write a function to demonstrate the behavior of mutable and immutable arguments.
def modify_data(a, b):
    a = a + 10          # Immutable (int)
    b.append(4)         # Mutable (list)
    print("Inside function:", a, b)

x = 5
y = [1, 2, 3]

modify_data(x, y)

print("Outside function:", x, y)
```

```python
#13. Write a Python program to find the sum of all even numbers between 1 and 50.
total = 0

for i in range(1, 51):
    if i % 2 == 0:
        total += i

print("Sum of even numbers:", total)
```

```python
#14. Write a program to reverse a string using a while loop.
text = input("Enter a string: ")
reverse = ""

i = len(text) - 1

while i >= 0:
    reverse += text[i]
    i -= 1

print("Reversed string:", reverse)
```

```python
#15. Write a Python program to calculate the factorial of a number provided by the user using a while loop.
num = int(input("Enter a number: "))
factorial = 1

i = 1
while i <= num:
    factorial *= i
    i += 1

print("Factorial:", factorial)
```
